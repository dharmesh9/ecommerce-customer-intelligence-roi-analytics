```python
# Cleaning for last_name
df['last_name'] = df['last_name'].str.strip().str.title()
df.loc[df['last_name'] == '', 'last_name'] = np.nan
```

```python
# EDA
print("\n=== email EDA ===")
print(f"Missing: {df['email'].isnull().sum()}")
print("Sample values:")
print(df['email'].head(10))
# Basic email regex
email_pat = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
invalid_email = df[~df['email'].str.match(email_pat, na=False)]
print(f"Invalid email count: {len(invalid_email)}")
print(invalid_email['email'].head(10))
```

```python
# Cleaning
df['email'] = df['email'].str.strip().str.lower()
# Optionally mark invalid emails
df['email_invalid'] = ~df['email'].str.match(email_pat, na=False)
# Replace empty strings with NaN
df.loc[df['email'] == '', 'email'] = np.nan
```

```python
# EDA
print("\n=== city EDA ===")
print(f"Missing: {df['city'].isnull().sum()} ({df['city'].isnull().mean()*100:.2f}%)")
print("Top 15 cities:")
print(df['city'].value_counts().head(15))
print("Sample unique values (may have casing/whitespace issues):")
print(df['city'].dropna().unique()[:20])
```

```python
# Cleaning
df['city'] = df['city'].str.strip().str.title()
df.loc[df['city'] == '', 'city'] = np.nan
most_common_city = df['city'].mode()[0]
df['city'] = df['city'].fillna(most_common_city)

# === INSERT THIS NEW CODE HERE ===
# Fix city typos and fill missing region by city
city_typo_map = {
    'Londn': 'London',
    'Londen': 'London',
    'Manchestr': 'Manchester',
    'Manchestor': 'Manchester',
    'Birmigham': 'Birmingham',
    'Brumingham': 'Birmingham',
    'Leds': 'Leeds',
    'Leedes': 'Leeds',
    'Bristoll': 'Bristol',
    'Liverpol': 'Liverpool',
    'Shefield': 'Sheffield',
    'Newcastel': 'Newcastle',
    'Cardif': 'Cardiff',
    'Edinburg': 'Edinburgh',
    'Edinbugh': 'Edinburgh',
    'Glasgoe': 'Glasgow',
    'Notingham': 'Nottingham',
}

print("\n=== Fixing city typos ===")
df['city'] = df['city'].map(city_typo_map).fillna(df['city'])
print(f"City values after typo fix: {df['city'].nunique()} unique cities")

# Create city to region mapping from existing data
city_region_map = df[df['region'].notna()].groupby('city')['region'].first().to_dict()

# Fill missing region by city
df['region'] = df.apply(
    lambda row: row['region'] if pd.notna(row['region']) else city_region_map.get(row['city'], row['region']),
    axis=1
)

print(f"Missing region after city-based fill: {df['region'].isnull().sum()}")
# ================================

# EDA
print("\n=== region EDA ===")
print(f"Missing: {df['region'].isnull().sum()} ({df['region'].isnull().mean()*100:.2f}%)")
print("Value counts:")
print(df['region'].value_counts())
```

```python
# EDA
print("\n=== region EDA ===")
print(f"Missing: {df['region'].isnull().sum()} ({df['region'].isnull().mean()*100:.2f}%)")
print("Value counts:")
print(df['region'].value_counts())
```

```python
# Cleaning
df['region'] = df['region'].str.strip().str.title()
df.loc[df['region'] == '', 'region'] = np.nan
# Optional: fill missing with mode
# most_common_region = df['region'].mode()[0]
# df['region'] = df['region'].fillna(most_common_region)
```

```python
# EDA
print("\n=== age EDA ===")
print(f"Missing: {df['age'].isnull().sum()} ({df['age'].isnull().mean()*100:.2f}%)")
print("Stats:")
print(df['age'].describe())
print("Plausibility check (age < 0 or age > 120):")
bad_age = df[(df['age'] < 0) | (df['age'] > 120)]
print(f"Count: {len(bad_age)}")
if len(bad_age) > 0:
    print(bad_age[['customer_id', 'age']].head(10))
```

```python
# === AGE CLEANING by GENDER ===

# First, set implausible ages to NaN
print("Setting implausible ages (<0 or >120) to NaN...")
df.loc[(df['age'] < 0) | (df['age'] > 120), 'age'] = np.nan

# Calculate mean age by gender
print("\n=== Mean age by gender ===")
mean_age_by_gender = df.groupby('gender')['age'].mean()
print(mean_age_by_gender)

# Impute missing age with gender-specific mean
print("\nImputing missing age with gender-specific mean...")
df['age'] = df.apply(
    lambda row: row['age'] if pd.notna(row['age']) else mean_age_by_gender[row['gender']],
    axis=1
)

# Check if there are still missing ages (cases where gender is also missing)
print(f"\nMissing age after imputation: {df['age'].isnull().sum()}")

# For any remaining missing age (where gender is also missing), use overall median
overall_median_age = df['age'].median()
df['age'] = df['age'].fillna(overall_median_age)
print(f"Filled remaining with overall median age: {overall_median_age:.2f}")

# Final check
print("\n=== Final age stats ===")
print(df['age'].describe())
```

```python
# EDA
print("\n=== gender EDA ===")
print(f"Missing: {df['gender'].isnull().sum()} ({df['gender'].isnull().mean()*100:.2f}%)")
print("Raw value counts:")
print(df['gender'].value_counts(dropna=False))
```

```python
def standardize_gender(g):
    if pd.isna(g):
        return np.nan

    g = str(g).strip().upper()

    if g in ['F', 'FEMALE', 'WOMAN']:
        return 'FEMALE'
    elif g in ['M', 'MALE', 'MAN']:
        return 'MALE'
    else:
        return 'Other'

df['gender'] = df['gender'].apply(standardize_gender)

print(df['gender'].value_counts(dropna=False))
```

```python
# EDA
print("\n=== signup_date EDA ===")
print(f"Missing: {df['signup_date'].isnull().sum()}")
print("Sample values:")
print(df['signup_date'].head(10))
print("Unique formats (first 20):")
print(df['signup_date'].dropna().unique()[:20])
```

```python
# Cleaning: convert to datetime
df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
print("After conversion, missing due to invalid format:")
print(df['signup_date'].isnull().sum())
# Optionally extract year/month
df['signup_year'] = df['signup_date'].dt.year
df['signup_month'] = df['signup_date'].dt.month
```

```python
# EDA
print("\n=== cac EDA ===")
print(f"Missing: {df['cac'].isnull().sum()}")
print("Stats:")
print(df['cac'].describe())
print("Zero CAC count:")
print((df['cac'] == 0).sum())
# IQR outlier detection
q1 = df['cac'].quantile(0.25)
q3 = df['cac'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = df[(df['cac'] < lower) | (df['cac'] > upper)]
print(f"IQR outliers count: {len(outliers)}")
print("Top 10 highest CAC:")
print(df.nlargest(10, 'cac')[['customer_id', 'acquisition_channel', 'cac']])
```

```python
# Check CAC = 0 by channel (all channels)
print("=== CAC = 0 by acquisition_channel (ALL CHANNELS) ===")
zeros_by_channel = df[df['cac'] == 0]['acquisition_channel'].value_counts()
print(zeros_by_channel)

# Total zeros
total_zeros = (df['cac'] == 0).sum()
print(f"\nTotal rows with CAC = 0: {total_zeros}")

# Zeros as percentage of each channel
print("\n=== CAC = 0 % by channel ===")
channel_counts = df['acquisition_channel'].value_counts()
zeros_pct = (zeros_by_channel / channel_counts * 100).round(2)
print(zeros_pct)

# Show sample of zeros for each channel
print("\n=== Sample CAC = 0 rows by channel ===")
for channel in df['acquisition_channel'].unique():
    zeros_in_channel = df[(df['cac'] == 0) & (df['acquisition_channel'] == channel)]
    if len(zeros_in_channel) > 0:
        print(f"\n{channel}: {len(zeros_in_channel)} zeros")
        print(zeros_in_channel[['customer_id', 'acquisition_channel', 'cac']].head(5))
```

```python
# Outliers by channel (IQR method per channel)
print("=== CAC OUTLIERS BY CHANNEL ===\n")

outlier_summary = []

for channel in df['acquisition_channel'].unique():
    channel_data = df[df['acquisition_channel'] == channel]['cac']
    
    q1 = channel_data.quantile(0.25)
    q3 = channel_data.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    
    outliers = df[(df['acquisition_channel'] == channel) & 
                  ((df['cac'] < lower) | (df['cac'] > upper))]
    
    outlier_summary.append({
        'channel': channel,
        'count': len(channel_data),
        'mean_cac': channel_data.mean(),
        'median_cac': channel_data.median(),
        'q1': q1,
        'q3': q3,
        'iqr': iqr,
        'lower_bound': lower,
        'upper_bound': upper,
        'outlier_count': len(outliers),
        'outlier_pct': (len(outliers) / len(channel_data) * 100) if len(channel_data) > 0 else 0
    })

# Summary table
summary_df = pd.DataFrame(outlier_summary)
summary_df = summary_df.round(2)
print(summary_df[['channel', 'count', 'mean_cac', 'median_cac', 'lower_bound', 'upper_bound', 'outlier_count', 'outlier_pct']])

# Show outlier details for each channel
print("\n=== OUTLIER DETAILS BY CHANNEL ===")
for channel in df['acquisition_channel'].unique():
    channel_data = df[df['acquisition_channel'] == channel]['cac']
    q1 = channel_data.quantile(0.25)
    q3 = channel_data.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    
    outliers = df[(df['acquisition_channel'] == channel) & 
                  ((df['cac'] < lower) | (df['cac'] > upper))]
    
    if len(outliers) > 0:
        print(f"\n{channel}: {len(outliers)} outliers (bounds: {lower:.2f} - {upper:.2f})")
        print(outliers[['customer_id', 'acquisition_channel', 'cac']].sort_values('cac', ascending=False).head(10))
```

```python
# === CHECK MISSING REGION AND WHERE IT SHOULD BELONG ===

print("=== 1. Rows with missing region ===")
missing_region = df[df['region'].isnull()]
print(f"Total rows with missing region: {len(missing_region)}")
print(missing_region[['customer_id', 'city', 'region']].head(20))

print("\n=== 2. Unique cities with missing region ===")
cities_missing_region = missing_region['city'].dropna().unique()
print(f"Number of unique cities: {len(cities_missing_region)}")
print("Cities:")
for city in cities_missing_region[:20]:
    print(f"  - {city}")

print("\n=== 3. City-Region mapping from existing data ===")
# Get city-region pairs where both exist
city_region_pairs = df[['city', 'region']].dropna().groupby(['city', 'region']).size().reset_index(name='count')
print(f"Total unique city-region pairs: {len(city_region_pairs)}")
print(city_region_pairs.sort_values('count', ascending=False).head(30))

print("\n=== 4. Which cities have missing region AND have a known region in other rows ===")
# Cities that appear both with and without region
cities_with_region = df[df['region'].notna()]['city'].unique()
cities_without_region = cities_missing_region

overlap = [city for city in cities_without_region if city in cities_with_region]
no_overlap = [city for city in cities_without_region if city not in cities_with_region]

print(f"Cities that CAN be mapped (also appear with region elsewhere): {len(overlap)}")
if len(overlap) > 0:
    for city in overlap[:15]:
        region = df[df['city'] == city]['region'].first()
        print(f"  {city} -> {region}")

print(f"\nCities that CANNOT be mapped (never appear with region): {len(no_overlap)}")
if len(no_overlap) > 0:
    for city in no_overlap[:15]:
        print(f"  {city} (no region known)")
```

```python
# === FILL MISSING REGION BY CITY ===

# Create city to region mapping from existing data
city_region_map = df[df['region'].notna()].groupby('city')['region'].first().to_dict()

print("=== City to Region Mapping ===")
print(f"Total cities mapped: {len(city_region_map)}")
print("Sample mappings:")
for city, region in list(city_region_map.items())[:10]:
    print(f"  {city} -> {region}")

# Check if London is in the mapping
print(f"\nLondon in mapping: {'London' in city_region_map}")
if 'London' in city_region_map:
    print(f"London -> {city_region_map['London']}")

# Fill missing region
df['region'] = df.apply(
    lambda row: row['region'] if pd.notna(row['region']) else city_region_map.get(row['city'], row['region']),
    axis=1
)

# Final check
print("\n=== After filling region ===")
print(f"Missing region: {df['region'].isnull().sum()}")
print("\nRegion value counts:")
print(df['region'].value_counts())

# Show any remaining missing region
remaining_missing = df[df['region'].isnull()]
if len(remaining_missing) > 0:
    print(f"\n=== Remaining missing region (cannot be mapped): {len(remaining_missing)} ===")
    print(remaining_missing[['customer_id', 'city', 'region']].head(10))
    print(f"\nUnique cities with still missing region: {remaining_missing['city'].unique()}")
```

```python
print("\n=== POST-CLEANING INFO ===")
df.info()

print("\n=== MISSING VALUES AFTER CLEANING ===")
missing_final = df.isnull().sum()
missing_pct_final = (df.isnull().sum() / len(df) * 100).round(2)
pd.DataFrame({'Missing_Count': missing_final, 'Missing_Prct': missing_pct_final})

print("\n=== FINAL SHAPE ===")
print(f"Rows: {len(df)}, Columns: {df.shape[1]}")

# Save cleaned data
df.to_csv('customers_cleaned.csv', index=False)
print("\nCleaned data saved to customers_cleaned.csv")
```
