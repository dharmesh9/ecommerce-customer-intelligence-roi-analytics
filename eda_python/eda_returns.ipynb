```python
# Check product_id
print("product_id nulls:", df['product_id'].isna().sum())
print("product_id unique:", df['product_id'].nunique())
print("Total rows:", len(df))
print("First 5 product_id values:")
print(df['product_id'].head())

# Strip whitespace if any
df['product_id'] = df['product_id'].str.strip()

# Check if product_id is unique
print("\nAll unique?", df['product_id'].nunique() == len(df))

# Show product_id value counts (how many times each product appears)
print("\nTop 10 product_id counts:")
print(df['product_id'].value_counts().head(10))
```

```python
# Check product_name
print("product_name nulls:", df['product_name'].isna().sum())
print("product_name unique:", df['product_name'].nunique())
print("Total rows:", len(df))
print("First 5 product_name values:")
print(df['product_name'].head())

# Strip whitespace
df['product_name'] = df['product_name'].str.strip()

# Check if product_name matches product_id consistently
print("\nProduct name vs product_id mapping (top 10):")
print(df.groupby(['product_id', 'product_name']).size().head(10))

# Check if any product_id has multiple product names (inconsistency)
print("\nProduct IDs with multiple names:")
multiple_names = df.groupby('product_id')['product_name'].nunique()
print(multiple_names[multiple_names > 1])
```

```python
# Check supplier
print("supplier nulls:", df['supplier'].isna().sum())
print("supplier unique:", df['supplier'].nunique())
print("First 10 supplier values:")
print(df['supplier'].head(10))

# Check case variations
print("\nSupplier value counts (lowercased, top 15):")
print(df['supplier'].str.lower().value_counts().head(15))
```

```python
# Clean supplier names
df['supplier'] = df['supplier'].str.strip().str.lower()

supplier_map = {
    'techcore ltd': 'TechCore Ltd',
    'techcoreltd': 'TechCore Ltd',
    'techcore': 'TechCore Ltd',
    'tech core ltd': 'TechCore Ltd',
    'elitecomponents': 'EliteComponents',
    'elite components': 'EliteComponents',
    'novatech supply': 'NovaTech Supply',
    'nova tech supply': 'NovaTech Supply',
    'novatechsupply': 'NovaTech Supply',
    'value source co': 'ValueSource Co',
    'valuesource co': 'ValueSource Co',
    'valuesourceco': 'ValueSource Co',
    'value source': 'ValueSource Co',
    'valuesource': 'ValueSource Co',
    'brightgadgets': 'BrightGadgets',
    'bright gadgets': 'BrightGadgets'
}

df['supplier'] = df['supplier'].map(supplier_map).fillna(df['supplier'].str.title())

# Verify cleaning
print("Cleaned supplier value counts:")
print(df['supplier'].value_counts())
```

```python
# Fix remaining Novatech variation
df['supplier'] = df['supplier'].replace('Novatech', 'NovaTech Supply')

# Verify final cleaning
print("Final supplier value counts:")
print(df['supplier'].value_counts())
print("\nUnique suppliers:", df['supplier'].nunique())
```

```python
# Check category
print("category nulls:", df['category'].isna().sum())
print("category unique:", df['category'].nunique())
print("First 10 category values:")
print(df['category'].head(10))

# Show value counts (excluding NaN)
print("\nCategory value counts:")
print(df['category'].value_counts(dropna=True))
```

```python
print(df['category'].to_string(index=False))
```

```python
# Show all rows with missing category
missing_category = df[df['category'].isna()]
print("Rows with missing category:", len(missing_category))
print("\nColumns to show:")
print(missing_category[['return_id', 'order_id', 'product_id', 'product_name', 'supplier', 'category', 'return_reason']])
```

```python
# Create product_id -> category mapping from rows where category exists
product_category_map = df[df['category'].notna()][['product_id', 'category']].drop_duplicates()
product_category_map = product_category_map.set_index('product_id')['category'].to_dict()

print("Product to Category mapping:")
for pid, cat in sorted(product_category_map.items()):
    print(f"  {pid}: {cat}")

# Fill missing category using the mapping
df['category'] = df.apply(
    lambda row: product_category_map[row['product_id']] if pd.isna(row['category']) else row['category'],
    axis=1
)

# Verify
print("\ncategory nulls after fill:", df['category'].isna().sum())
print("\nCategory value counts:")
print(df['category'].value_counts())
```

```python
# Check return_date
print("return_date nulls:", df['return_date'].isna().sum())
print("First 5 return_date values:")
print(df['return_date'].head())

# Convert to datetime
df['return_date'] = pd.to_datetime(df['return_date'], errors='coerce')

# Check how many failed to parse
print("\nFailed to parse:", df['return_date'].isna().sum())

# Show date range
print("Date range:", df['return_date'].min(), "to", df['return_date'].max())

# Show distribution by year
print("\nReturns by year:")
print(df['return_date'].dt.year.value_counts().sort_index())
```

```python
# Check return_reason
print("return_reason nulls:", df['return_reason'].isna().sum())
print("return_reason unique:", df['return_reason'].nunique())
print("First 10 return_reason values:")
print(df['return_reason'].head(10))

# Show value counts (excluding NaN)
print("\nReturn reason value counts:")
print(df['return_reason'].value_counts(dropna=True))
```

```python
# Show all rows with missing return_reason
missing_reason = df[df['return_reason'].isna()]
print("Rows with missing return_reason:", len(missing_reason))
print("\nColumns to show:")
print(missing_reason[['return_id', 'order_id', 'product_id', 'product_name', 'supplier', 'category', 'return_reason', 'refund_amount']])
```

```python
# Create product_id -> most common return_reason mapping
product_reason_map = df[df['return_reason'].notna()].groupby('product_id')['return_reason'].agg(
    lambda x: x.value_counts().index[0]
).to_dict()

print("Product to most common return reason mapping:")
for pid, reason in sorted(product_reason_map.items()):
    print(f"  {pid}: {reason}")

# Fill missing return_reason using the mapping
df['return_reason'] = df.apply(
    lambda row: product_reason_map[row['product_id']] if pd.isna(row['return_reason']) else row['return_reason'],
    axis=1
)

# Verify
print("\nreturn_reason nulls after fill:", df['return_reason'].isna().sum())
print("\nReturn reason value counts:")
print(df['return_reason'].value_counts())
```

```python
# Check refund_amount
print("refund_amount nulls:", df['refund_amount'].isna().sum())
print("refund_amount describe:")
print(df['refund_amount'].describe())

# Check for negative or zero values
print("\nNegative refunds:", (df['refund_amount'] < 0).sum())
print("Zero refunds:", (df['refund_amount'] == 0).sum())

# Check for outliers using IQR
Q1 = df['refund_amount'].quantile(0.25)
Q3 = df['refund_amount'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['refund_amount'] < lower) | (df['refund_amount'] > upper)]
print("\nOutliers (IQR method):", outliers.shape[0])
print("Outlier values:")
print(outliers[['product_id', 'product_name', 'refund_amount', 'qty_returned']].head(10))

# Show max refund amounts
print("\nTop 10 highest refund amounts:")
print(df.nlargest(10, 'refund_amount')[['product_id', 'product_name', 'refund_amount', 'qty_returned']])
```

```python
# Show negative refund rows
print("Negative refund rows:")
print(df[df['refund_amount'] < 0][['return_id', 'product_id', 'product_name', 'refund_amount', 'qty_returned', 'return_reason']])

# Check if high refunds match high-priced products
print("\nCheck refund per unit (refund_amount / qty_returned):")
df['refund_per_unit'] = df['refund_amount'] / df['qty_returned']
print(df.nlargest(10, 'refund_per_unit')[['product_id', 'product_name', 'refund_amount', 'qty_returned', 'refund_per_unit']])
```

```python
# Fix negative refunds (make absolute value)
df.loc[df['refund_amount'] < 0, 'refund_amount'] = df.loc[df['refund_amount'] < 0, 'refund_amount'].abs()

# Verify no more negative refunds
print("Negative refunds after fix:", (df['refund_amount'] < 0).sum())

# Check refund_per_unit for potential data errors
df['refund_per_unit'] = df['refund_amount'] / df['qty_returned']

# Flag suspicious high refund_per_unit (e.g., > 500 for most products)
print("\nSuspicious high refund_per_unit (>500):")
suspicious = df[df['refund_per_unit'] > 500][['return_id', 'product_id', 'product_name', 'refund_amount', 'qty_returned', 'refund_per_unit']]
print(suspicious)

# Optional: Flag these rows (add a column)
df['suspicious_refund'] = df['refund_per_unit'] > 500
print("\nSuspicious refund count:", df['suspicious_refund'].sum())
```

```python
# Check qty_returned
print("qty_returned nulls:", df['qty_returned'].isna().sum())
print("qty_returned describe:")
print(df['qty_returned'].describe())

# Check for negative or zero values
print("\nNegative qty:", (df['qty_returned'] < 0).sum())
print("Zero qty:", (df['qty_returned'] == 0).sum())

# Show unique values
print("\nUnique qty values:")
print(df['qty_returned'].value_counts().sort_index())

# Check max quantities
print("\nTop 10 highest qty_returned:")
print(df.nlargest(10, 'qty_returned')[['return_id', 'product_id', 'product_name', 'qty_returned', 'refund_amount']])
```

```python
# Fix the 999 qty error (likely a typo — set to 1 or investigate)
print("Row with qty=999:")
print(df[df['qty_returned'] == 999][['return_id', 'product_id', 'product_name', 'qty_returned', 'refund_amount']])

# Fix: set 999 to 1 (most likely intended value)
df.loc[df['qty_returned'] == 999, 'qty_returned'] = 1

# Verify
print("\nqty_returned describe after fix:")
print(df['qty_returned'].describe())
print("\nUnique qty values after fix:")
print(df['qty_returned'].value_counts().sort_index())
```
