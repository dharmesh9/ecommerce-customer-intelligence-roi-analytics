```python
product_summary = df.groupby("product_id").agg(
    avg_stock_level=("stock_level", "mean"),
    avg_units_sold=("units_sold", "mean"),
    stockout_rate=("stockout_flag", "mean")
).reset_index()
```

```python
product_summary.sort_values("stockout_rate", ascending=False)
```
