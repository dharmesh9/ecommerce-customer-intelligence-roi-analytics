```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('date_dim.csv')
```

```python
df.info()
```

```python
df.head(30)
```

```python
# Convert to datetime
df['date'] = pd.to_datetime(df['date'])

# Check results
print(f"Data type: {df['date'].dtype}")
print(f"Non-null count: {df['date'].notnull().sum()}")
print(f"Min date: {df['date'].min()}")
print(f"Max date: {df['date'].max()}")
print(f"First 5 values:\n{df['date'].head()}")
```

```python
# Verify year matches date
year_from_date = df['date'].dt.year
year_mismatch = df['year'] != year_from_date

print(f"Data type: {df['year'].dtype}")
print(f"Non-null count: {df['year'].notnull().sum()}")
print(f"Unique years: {df['year'].unique()}")
print(f"Year value counts:\n{df['year'].value_counts().sort_index()}")
print(f"\nYears mismatching date: {year_mismatch.sum()}")

if year_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[year_mismatch][['date', 'year']])
else:
    print("\n✓ All years match the date column!")
```

```python
#Verify quarter matches date
quarter_from_date = df['date'].dt.quarter.map(lambda x: f'Q{x}')
quarter_mismatch = df['quarter'] != quarter_from_date

print(f"Data type: {df['quarter'].dtype}")
print(f"Non-null count: {df['quarter'].notnull().sum()}")
print(f"Unique quarters: {df['quarter'].unique()}")
print(f"\nQuarter value counts:\n{df['quarter'].value_counts()}")
print(f"\nQuarters mismatching date: {quarter_mismatch.sum()}")

if quarter_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[quarter_mismatch][['date', 'quarter']].head(10))
else:
    print("\n✓ All quarters match the date column!")
```

```python
# Verify month_num matches date
month_from_date = df['date'].dt.month
month_mismatch = df['month_num'] != month_from_date

print(f"Data type: {df['month_num'].dtype}")
print(f"Non-null count: {df['month_num'].notnull().sum()}")
print(f"Unique values: {sorted(df['month_num'].unique())}")
print(f"\nMonth number value counts:\n{df['month_num'].value_counts().sort_index()}")
print(f"\nMonths mismatching date: {month_mismatch.sum()}")

if month_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[month_mismatch][['date', 'month_num']].head(10))
else:
    print("\n✓ All month_num values match the date column!")
```

```python
# Verify month_name matches date
month_name_from_date = df['date'].dt.month_name()
month_name_mismatch = df['month_name'] != month_name_from_date

print(f"Data type: {df['month_name'].dtype}")
print(f"Non-null count: {df['month_name'].notnull().sum()}")
print(f"Unique values: {sorted(df['month_name'].unique())}")
print(f"\nMonth name value counts:\n{df['month_name'].value_counts()}")
print(f"\nMonth names mismatching date: {month_name_mismatch.sum()}")

if month_name_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[month_name_mismatch][['date', 'month_name']].head(10))
else:
    print("\n✓ All month_name values match the date column!")
```

```python
# Verify week_num matches ISO week number
week_from_date = df['date'].dt.isocalendar().week
week_mismatch = df['week_num'] != week_from_date

print(f"Data type: {df['week_num'].dtype}")
print(f"Non-null count: {df['week_num'].notnull().sum()}")
print(f"Week range: {df['week_num'].min()} to {df['week_num'].max()}")
print(f"\nWeek number value counts (first 10):\n{df['week_num'].value_counts().sort_index().head(10)}")
print(f"\nWeeks mismatching date: {week_mismatch.sum()}")

if week_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS (first 10):")
    print(df[week_mismatch][['date', 'week_num']].head(10))
else:
    print("\n✓ All week_num values match the date column!")
```

```python
# Verify day_of_week matches date
day_from_date = df['date'].dt.day_name()
day_mismatch = df['day_of_week'] != day_from_date

print(f"Data type: {df['day_of_week'].dtype}")
print(f"Non-null count: {df['day_of_week'].notnull().sum()}")
print(f"Unique values: {sorted(df['day_of_week'].unique())}")
print(f"\nDay of week value counts:\n{df['day_of_week'].value_counts()}")
print(f"\nDays mismatching date: {day_mismatch.sum()}")

if day_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[day_mismatch][['date', 'day_of_week']].head(10))
else:
    print("\n✓ All day_of_week values match the date column!")
```

```python
# Verify is_weekend matches date (Saturday=5, Sunday=6)
is_weekend_verified = df['date'].dt.dayofweek >= 5
is_weekend_mismatch = df['is_weekend'] != is_weekend_verified

print(f"Data type: {df['is_weekend'].dtype}")
print(f"Non-null count: {df['is_weekend'].notnull().sum()}")
print(f"Value counts:\n{df['is_weekend'].value_counts()}")
print(f"\nTrue count (weekend): {df['is_weekend'].sum()}")
print(f"False count (weekday): {len(df) - df['is_weekend'].sum()}")
print(f"\nValues mismatching date: {is_weekend_mismatch.sum()}")

if is_weekend_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[is_weekend_mismatch][['date', 'day_of_week', 'is_weekend']].head(10))
else:
    print("\n✓ All is_weekend values match the date column!")
```

```python
# Verify is_month_end matches date
is_month_end_verified = df['date'].dt.is_month_end
is_month_end_mismatch = df['is_month_end'] != is_month_end_verified

print(f"Data type: {df['is_month_end'].dtype}")
print(f"Non-null count: {df['is_month_end'].notnull().sum()}")
print(f"Value counts:\n{df['is_month_end'].value_counts()}")
print(f"\nTrue count (month end): {df['is_month_end'].sum()}")
print(f"False count: {len(df) - df['is_month_end'].sum()}")
print(f"\nMonths end mismatching date: {is_month_end_mismatch.sum()}")

if is_month_end_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[is_month_end_mismatch][['date', 'is_month_end']].head(10))
else:
    print("\n✓ All is_month_end values match the date column!")
```

```python
# Fix: Standard pandas is_month_end (only last day of month)
df['is_month_end'] = df['date'].dt.is_month_end

print(f"Fixed is_month_end value counts:\n{df['is_month_end'].value_counts()}")
print(f"True count (month end): {df['is_month_end'].sum()}")
print(f"\nLast day of each month (first 12):")
print(df[df['is_month_end'] == True][['date', 'month_name']].head(12))
```

```python
print(f"Data type: {df['is_peak_season'].dtype}")
print(f"Non-null count: {df['is_peak_season'].notnull().sum()}")
print(f"Value counts:\n{df['is_peak_season'].value_counts()}")
print(f"\nTrue count (peak season): {df['is_peak_season'].sum()}")
print(f"False count: {len(df) - df['is_peak_season'].sum()}")

if df['is_peak_season'].sum() > 0:
    print(f"\nPeak season by month:")
    print(df[df['is_peak_season'] == True]['month_name'].value_counts())
    print(f"\nPeak season by season:")
    print(df[df['is_peak_season'] == True]['season'].value_counts())
else:
    print("\nNote: No peak season days marked in this dataset.")
    print("You may need to define which months/seasons are 'peak' for your business.")
```

```python
# Define expected season based on month
def get_expected_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:  # 9, 10, 11
        return 'Autumn'

season_verified = df['month_num'].apply(get_expected_season)
season_mismatch = df['season'] != season_verified

print(f"Data type: {df['season'].dtype}")
print(f"Non-null count: {df['season'].notnull().sum()}")
print(f"Unique values: {sorted(df['season'].unique())}")
print(f"\nSeason value counts:\n{df['season'].value_counts()}")
print(f"\nSeasons mismatching date: {season_mismatch.sum()}")

if season_mismatch.sum() > 0:
    print("\nMISMATCHED ROWS:")
    print(df[season_mismatch][['date', 'month_num', 'season']].head(10))
else:
    print("\n✓ All season values match the month_num column!")
```
