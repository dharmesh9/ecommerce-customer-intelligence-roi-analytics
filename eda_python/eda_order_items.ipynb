```python
# Fix cost_price: replace 0 values with product's median cost (safer version)

# Step 1: Calculate median cost for each product (excluding 0 values)
median_cost_by_product = df[df['cost_price'] > 0].groupby('product_id')['cost_price'].median()
print("=== Products with NO valid cost data ===")
missing_products = set(df['product_id'].unique()) - set(median_cost_by_product.index)
print(f"Products missing: {missing_products}")

# Step 2: Add a global median for products with no cost data
global_median_cost = df[df['cost_price'] > 0]['cost_price'].median()
median_cost_by_product = median_cost_by_product.fillna(global_median_cost)
print(f"\nGlobal median cost: {global_median_cost:.2f}")

# Step 3: Create a mapping dictionary
cost_map = median_cost_by_product.to_dict()

# Step 4: Replace 0 values with product's median cost (using .get() to avoid KeyError)
df['cost_price'] = df['cost_price'].apply(
    lambda x: cost_map.get(df.loc[df['cost_price'] == x].index[0] if x == 0 else -1, global_median_cost) if x == 0 else x
)

# Simpler approach - use map directly
df['cost_price'] = df.apply(
    lambda row: cost_map.get(row['product_id'], global_median_cost) if row['cost_price'] == 0 else row['cost_price'],
    axis=1
)

# Step 5: Verify the fix
print("\n=== Verification After Fix ===")
print(f"Zero values after fix: {(df['cost_price'] == 0).sum()}")
print(f"Null values after fix: {df['cost_price'].isna().sum()}")

# Step 6: New statistics
print("\n=== New Cost Price Statistics ===")
print(df['cost_price'].describe())
```
